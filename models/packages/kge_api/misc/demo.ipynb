```python
import json
import logging

import requests
```

```python
# The URL of your predict endpoint
predict_url = "http://localhost:8088/test/kge/predict"
score_url = "http://localhost:8088/test/kge/score"
score_index_url = "http://localhost:8088/test/kge/score_from_index"
fit_score_scaler_url = "http://localhost:8088/test/kge/fit_score_scaler"
device_url = "http://localhost:8088/test/kge/device"

SUCCESS_CODE = 200
# The data to send in the request


def create_request(url, data):
    # Convert the data to JSON
    json_data = json.dumps(data)

    # Set the headers
    headers = {"Content-Type": "application/json"}

    # Make the POST request
    response = requests.post(url, data=json_data, headers=headers)

    # Check the response
    if response.status_code == SUCCESS_CODE:
        # Request was successful
        result = response.json()
        logging.info(f"Prediction result: {result}")
        return result
    # There was an error
    print(f"Error: {response.status_code} {response.text}")
    return None
```
