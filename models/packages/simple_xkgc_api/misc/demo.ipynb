```python
import json
import logging

import requests
```

```python
# The URL of your predict endpoint
explain_url = "http://localhost:8089/test/path_explainer/explain"
explanation_len_url = "http://localhost:8089/test/path_explainer/explanation_length"

SUCCESS_CODE = 200
# The data to send in the request

def call_api(url, data):
    # Convert the data to JSON
    json_data = json.dumps(data)

    # Set the headers
    headers = {
        "Content-Type": "application/json"
    }

    # Make the POST request
    response = requests.post(url, data=json_data, headers=headers)

    # Check the response
    if response.status_code == SUCCESS_CODE:
        # Request was successful
        result = response.json()
        logging.info(f"Prediction result: {result}")
        return result
    # There was an error
    print(f"Error: {response.status_code} {response.text}")
    return None
```

```python
# Digital Science
triple = ["101000241533", "INHIBITS", "101000106424"]
# Pubtator
triple = ["00064ef166585da2e7dbe45b6826affe", "CAUSE", "00069bed05f3c40e3e0f8b05cbe694e9"]
triple = ["0d28423129e620ff4ef207fcf7df8b0d", "INHIBIT", "71d4034793735c89bdcee46cc8572747"]
```

```python

data = {
    "triples_to_probe": [triple],
    "num_explanations": 5,
    "explanation_configs": [],
    "rerank_strategy": "unique_pathways"}

# data["explanation_configs"].append({"type": "necessary", "batch_size": 8})
data["explanation_configs"].append({"type": "sufficient", "batch_size": 32})

output = call_api(explain_url, data)
explanations = output["explanations"]
print(f"Number of explanations: {len(explanations)}")
```
