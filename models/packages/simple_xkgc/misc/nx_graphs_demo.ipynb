```python
import networkx as nx
```

```python
# Create a MultiDiGraph
G = nx.MultiDiGraph()

# Add nodes
G.add_nodes_from(["A", "B"])

# Add edges with custom keys
G.add_edge("B", "A", key="relation_2", weight=10, color="blue")
G.add_edge("B", "A", key="relation_3", weight=15, color="green")
G.add_edge("B", "C", key="relation_1", weight=15, color="green")
G.add_edge("C", "B", key="relation_2", weight=15, color="green")
G.add_edge("C", "D", key="relation_3", weight=15, color="green")
```

```python
G_u = G.to_undirected()

for u, v, key, data in G_u.edges(data=True, keys=True):
    print(f"Edge ({u}, {v}, '{key}'): {data}")
```

```python
G.get_edge_data("A", "B")
```

```python
paths = list(
    nx.all_shortest_paths(
        G_u,
        source="A",
        target="D",
    )
)
paths
```

```python
for path in paths:
    triple_list = []

    for subject_i, object_i in zip(path[:-1], path[1:], strict=False):
        triple_list_i = []

        if G.has_edge(subject_i, object_i):
            relations = list(G.get_edge_data(subject_i, object_i).keys())
            for relation in relations:
                triple_list_i.append((subject_i, relation, object_i))
        if G.has_edge(object_i, subject_i):
            relations = list(G.get_edge_data(object_i, subject_i).keys())
            for relation in relations:
                triple_list_i.append((object_i, relation, subject_i))

        triple_list.append(triple_list_i)

triple_list
```

```python
import itertools

paths_triples = list(itertools.product(*triple_list))
```

```python
for path_i in paths_triples:
    print(path_i)
```
