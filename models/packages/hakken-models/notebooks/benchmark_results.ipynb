```python
import pandas as pd

from hakken_models.results.latex import format_df_for_latex
```

```python
ROOT_FOLDER = "/home/pablo.sanchez2/GitHub/project_spaice_ds/packages/pip/hakken-models"
results_file = f"{ROOT_FOLDER}/results/benchmark/1990/comparison_summary.csv"
results_df = pd.read_csv(results_file, sep=",")
```

```python
for i, c in enumerate(results_df.columns):
    print(f"[{i}] {c}")
```

```python
from hakken_models.results.entities import ColumnName, MetricInfo, ModelInfo, OriginalName

# --- Models (colors: tab10-style) ---
_MODEL_COLORS = ("#1f77b4", "#ff7f0e", "#2ca02c", "#d62728", "#9467bd", "#8c564b", "#e377c2", "#7f7f7f")
model_name_mapping: dict[OriginalName, ModelInfo] = {
    "Random": ModelInfo("Random", 1, _MODEL_COLORS[0]),
    "KGE-ComplEX": ModelInfo("ComplEx", 2, _MODEL_COLORS[1]),
    "KNN": ModelInfo("KNN", 3, _MODEL_COLORS[2]),
    "MLP": ModelInfo("MLP", 4, _MODEL_COLORS[3]),
    "Rule_Based": ModelInfo("RuleBased", 5, _MODEL_COLORS[4]),
    "Agatha": ModelInfo("Agatha", 6, _MODEL_COLORS[5]),
    "TNodeEmbed": ModelInfo("TNodeEmbed", 7, _MODEL_COLORS[6]),
    "ThiGER": ModelInfo("THiGER", 8, _MODEL_COLORS[7]),
}

# --- Metrics (refactored to use MetricInfo) ---
metric_mapping: dict[ColumnName, MetricInfo] = {
    # --- Overall averages ---
    "avg_macro_prec": MetricInfo(r"$\mathrm{Prec}_{\mathrm{macro}}$", True),
    "avg_macro_rec": MetricInfo(r"$\mathrm{Rec}_{\mathrm{macro}}$", True),
    "avg_macro_f1": MetricInfo(r"$F_{1,\mathrm{macro}}$", True),
    "avg_weighted_prec": MetricInfo(r"$\mathrm{Prec}_{\mathrm{w}}$", True),
    "avg_weighted_rec": MetricInfo(r"$\mathrm{Rec}_{\mathrm{w}}$", True),
    "avg_weighted_f1": MetricInfo(r"$F_{1,\mathrm{w}}$", True),
    # --- Masked up to K ---
    # "masked_upto_K_macro_prec": MetricInfo(
    #     r"$\mathrm{Prec}^{\mathrm{mask}\leq K}_{\mathrm{macro}}$", True
    # ),
    # "masked_upto_K_macro_rec": MetricInfo(r"$\mathrm{Rec}^{\mathrm{mask}\leq K}_{\mathrm{macro}}$", True),
    # "masked_upto_K_macro_f1": MetricInfo(r"$F_{1,\mathrm{macro}}^{\mathrm{mask}\leq K}$", True),
    # "masked_upto_K_weighted_prec": MetricInfo(r"$\mathrm{Prec}^{\mathrm{mask}\leq K}_{\mathrm{w}}$", True),
    # "masked_upto_K_weighted_rec": MetricInfo(r"$\mathrm{Rec}^{\mathrm{mask}\leq K}_{\mathrm{w}}$", True),
    # "masked_upto_K_weighted_f1": MetricInfo(r"$F_{1,\mathrm{w}}^{\mathrm{mask}\leq K}$", True),
    # --- Future predictions at K ---
    # "future_predictions_at_K_macro_prec": MetricInfo(
    #     r"$\mathrm{Prec}^{\mathrm{future}@K}_{\mathrm{macro}}$", True
    # ),
    # "future_predictions_at_K_macro_rec": MetricInfo(r"$\mathrm{Rec}^{\mathrm{future}@K}_{\mathrm{macro}}$", True),
    # "future_predictions_at_K_macro_f1": MetricInfo(r"$F_{1,\mathrm{macro}}^{\mathrm{future}@K}$", True),
    # "future_predictions_at_K_weighted_prec": MetricInfo(r"$\mathrm{Prec}^{\mathrm{future}@K}_{\mathrm{w}}$", True),
    # "future_predictions_at_K_weighted_rec": MetricInfo(r"$\mathrm{Rec}^{\mathrm{future}@K}_{\mathrm{w}}$", True),
    # "future_predictions_at_K_weighted_f1": MetricInfo(r"$F_{1,\mathrm{w}}^{\mathrm{future}@K}$", True),
    # --- Ranking metrics ---
    # "ndcg_1": MetricInfo(r"$\mathrm{nDCG}@1$", True),
    # "ndcg_3": MetricInfo(r"$\mathrm{nDCG}@3$", True),
    # "ndcg_5": MetricInfo(r"$\mathrm{nDCG}@5$", True),
    # "ndcg_10": MetricInfo(r"$\mathrm{nDCG}@10$", True),
    "ndcg_all_m": MetricInfo(r"$\mathrm{nDCG}_{\mathrm{mean}}$", True),
}

# Optional: derive your old "columns_mapping" (display-only) from MetricInfo
columns_mapping: dict[ColumnName, ColumnName] = {}
columns_mapping["model"] = r"\textbf{Model}"

columns_mapping.update({k: v.display_name for k, v in metric_mapping.items()})
```

```python
cols_to_keep = list(columns_mapping)
latex_df = results_df[cols_to_keep].copy()

latex_df = format_df_for_latex(
    latex_df,
    metric_mapping=metric_mapping,
    decimals=2,
    multiplier=100.0,
    rename_cols=columns_mapping,  # your latex column names
    model_name_mapping=model_name_mapping,  # optional
    format_metrics=True,
)


# latex_df = latex_df.rename(columns=columns_mapping)

latex_df
```

```python
print(latex_df.to_latex(index=False, escape=False))
```

```python
plot_df = format_df_for_latex(
    results_df[cols_to_keep].copy(),
    metric_mapping=metric_mapping,
    decimals=2,
    rename_cols=columns_mapping,  # your latex column names
    model_name_mapping=model_name_mapping,  # optional
    format_metrics=False,
)

plot_df.fillna(0.0, inplace=True)
plot_df
```

```python
import matplotlib.pyplot as plt

from hakken_models.results.entities import LegendConfig, SaveFigureConfig
from hakken_models.results.plots import mpl_style, plot_radar_chart

legend_config = LegendConfig(
    loc="upper right", bbox_to_anchor=(1.25, 1.15), frameon=False, ncol=plot_df.shape[0]
)

plot_file = ROOT_FOLDER + "/results/benchmark/benchmark_radar.pdf"
legend_plot_file = ROOT_FOLDER + "/results/benchmark/benchmark_radar_legend.pdf"
with mpl_style("nature"):
    fig, ax = plot_radar_chart(
        plot_df,
        model_col=columns_mapping.get("model", "model"),
        normalize=False,
        remove_legend=True,
        title="",
        legend_config=legend_config,
        save_config=SaveFigureConfig(path=plot_file, dpi=120, bbox_inches="tight"),
        save_legend_config=SaveFigureConfig(path=legend_plot_file, dpi=120, bbox_inches="tight"),
    )

    plt.show()
```
