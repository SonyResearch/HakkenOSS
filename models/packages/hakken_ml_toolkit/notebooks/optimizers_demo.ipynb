```python
import torch

from hakken_ml_toolkit.optimizers import AdamConfig, OptimizerFactory, OptimizerType
```

```python
# Create a model
num_features = 10
model = torch.nn.Linear(num_features, 1)

# Configure and create the optimizer
config = AdamConfig(learning_rate=0.01, betas=(0.9, 0.999))
optimizer = OptimizerFactory.create(
    model.parameters(), optim_type=OptimizerType.ADAM, config=config
)
```

```python
# Use in training loop
num_epochs = 10
batch_size = 16
for _ in range(num_epochs):
    x = torch.randn(batch_size, num_features)
    loss_all: torch.FloatTensor = model(x)
    loss_mean = loss_all.mean()
    loss_mean.backward()
    optimizer.step()
    optimizer.zero_grad()
    print(f"loss: {loss_mean.item():.2f}")
```
