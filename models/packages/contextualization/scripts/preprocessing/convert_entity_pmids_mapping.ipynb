A simple script for converting `entity_pmids_mapping.json` into the ndjson format.

```python
from pathlib import Path

MAPPING_JSON_PATH = Path("~/data/pubtator3-v0.4.0/entity_pmids_mapping.json").expanduser()
```

```python
import json

with open(MAPPING_JSON_PATH) as f:
    mapping = json.load(f)
```

```python
SAVE_DIR = Path("~/data/pubtator3-v0.4.0/entity_pmids/").expanduser()
SAVE_DIR.mkdir(exist_ok=True, parents=True)
```

```python
MAX_LINES_PER_FILE = 500000
```

```python
from tqdm.auto import tqdm
```

```python
buffer = []
file_index = 0

for node_id, pmids in tqdm(mapping.items()):
    for pmid in pmids:
        buffer.append({"node_id": str(node_id), "pmid": str(pmid)})
        if len(buffer) >= MAX_LINES_PER_FILE:
            filename = f"{file_index:04d}.ndjson"
            file_path = SAVE_DIR / filename

            with open(file_path, "w") as f:
                for obj in buffer:
                    f.write(json.dumps(obj))
                    f.write("\n")
            file_index += 1
            buffer = []

if len(buffer) > 0:
    filename = f"{file_index:04d}.ndjson"
    file_path = SAVE_DIR / filename
    with open(file_path, "w") as f:
        for obj in buffer:
            f.write(json.dumps(obj))
            f.write("\n")
    file_index += 1
    buffer = []
```
