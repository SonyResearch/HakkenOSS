```python
from kge.models import KGEAPI
from hakken_ml_toolkit.ml_utils.extras import TensorCreator
```

```python
# Initialize the model with connection to your local KGE API server
model = KGEAPI(base_url="http://localhost:8088/test/kge")
```

```python
# Create a sample batch of data using TensorCreator
sro_batch = TensorCreator.long_tensor([[0, 0, 2], [1, 2, 2]], device="cuda")
print("Sample Input Batch:")
print(sro_batch)
```

```python
# Score the sample batch using the model
scores = model.score(sro_batch=sro_batch)
print("\nModel Scores:")
print(scores)
```
