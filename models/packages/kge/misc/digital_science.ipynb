```python
from dotenv import load_dotenv
import os

import pandas as pd
load_dotenv()
```

```python
folder = os.getenv("DATA_PATH")

file_path = os.path.join(folder, "edges.csv")
df = pd.read_csv(file_path, sep="\t", header=0)

df.head(2)
```

```python
num_rows = df.shape[0]

df_cleaned = df.dropna()

ratio = df_cleaned.shape[0]/num_rows
print(f"{df.shape} {df_cleaned.shape} {ratio*100:2f}")
```

```python
from datetime import datetime
df_cleaned["year"] = df_cleaned.timestamp.apply(lambda x: datetime.fromtimestamp(x).year)
```

```python
cond =(df_cleaned.year < 2003) & (df_cleaned.year > 1970)
small_df = df_cleaned[cond]
small_df.shape
```
